```python
import pandas as pd

candidates_df = pd.read_csv(CANDIDATE_PATH, sep="\t")
gt = pd.read_csv(GROUND_TRUTH_PATH  , sep="\t")

rows = []
gt_dict = dict(zip(gt['source1_entity_id'], gt['matched_entity_ids'].fillna('')))

for _, row in candidates_df.iterrows():
    s1_id = row['source1_entity_id']
    true_matches = set(gt_dict.get(s1_id, '').split(',')) if gt_dict.get(s1_id, '') else set()
    for cand_id in str(row['candidate_entity_ids']).split(','):
        if cand_id == '' or cand_id == 'nan':
            continue
        rows.append({
            'source1_entity_id': s1_id,
            'candidate_entity_id': cand_id,
            'is_match': 1 if cand_id in true_matches else 0
        })

pairs_df = pd.DataFrame(rows)
print(pairs_df['is_match'].value_counts())  # imbalance check karo
```
